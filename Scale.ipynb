```python
import pandas as pd
import numpy as np
```

```python
def normalize(columns):
    col = columns.copy()
    min_val = col.min()
    max_val = col.max()
    if max_val == min_val:
        return pd.Series(0.0, index = col.index)
    return (col - min_val) / (max_val - min_val)

def gaming_score(df):
    weights = {'antutu_11': 0.5, 'RAM': 0.3, 'Refresh Rate': 0.2}
    
    norm = pd.DataFrame(index = df.index)
    norm['antutu_11']    = normalize(df['antutu_11'])
    norm['RAM']          = normalize(df['RAM'])
    norm['Refresh Rate'] = normalize(df['Refresh Rate'])  # cap PWM outliers
 
    score = sum(norm[col] * w for col, w in weights.items())
    return score.round(3)

def compute_photo_score(df):
    weights = {
        'antutu_11'     : 0.25,
        'rear_ois'      : 0.25,
        'rear_mp_max'   : 0.20,
        'rear_wide'     : 0.20,
        'rear_telephoto': 0.05,
        'front_mp'      : 0.05,
    }
 
    norm = pd.DataFrame(index=df.index)
    norm['antutu_11']      = normalize(df['antutu_11'])
    norm['rear_ois']       = df['rear_ois'].astype(float)       
    norm['rear_mp_max']    = normalize(df['rear_mp_max'])
    norm['rear_wide']      = df['rear_wide'].astype(float)     
    norm['rear_telephoto'] = df['rear_telephoto'].astype(float)
    norm['front_mp']       = normalize(df['front_mp'])
 
    score = sum(norm[col] * w for col, w in weights.items())
    return score.round(3)
    
```
